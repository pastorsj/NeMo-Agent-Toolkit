## Plot Charts
Here we can use the Python SDK to develop a chart plotting workflow, then save it to a config.yaml and run it from there.

This workflow creates charts (line, bar, or scatter plots) from data based on user requests.


```python
import os
import sys

# Import the NeMo-Agent-Toolkit module
module_path = os.path.abspath('../../../src/')
if module_path not in sys.path:
    sys.path.insert(0, module_path)
```

```python
import logging

logger = logging.getLogger()
logger.setLevel(logging.DEBUG)
```

```python
from nat.llm.sdk import NimLLM
from nat.utils.sdk.nat_workflow import NatWorkflow
from nat_plot_charts.sdk import PlotChartsWorkflow

llm = NimLLM(
    model_name="nvdev/meta/llama-3.1-70b-instruct",
    temperature=0.0,
    name="nim_llm",
)

# Create the plot charts workflow
# Note: data_file_path should point to example_data.json which has the xValues/yValues structure
workflow = PlotChartsWorkflow(
    llm=llm,
    data_file_path="../example_data.json",
    output_directory="outputs",
    chart_types=["line", "bar", "scatter"],
    max_data_points=100,
    figure_size=(10, 6),
)

nat_workflow = NatWorkflow(
    entrypoint=workflow,
)
```

```python
import os
from pathlib import Path

path_to_yaml = Path(os.getcwd(), "config", "config.yaml").resolve()

# Create the config directory if it doesn't exist
if not path_to_yaml.parent.exists():
    os.makedirs(path_to_yaml.parent)

# Save the workflow to a config file
nat_workflow.save_to_config_file(path_to_yaml)

# Print out the config file content
with open(path_to_yaml) as f:
    print(f.read())
```

```python
# Create a line chart from the data
await nat_workflow.prompt("Create a line chart from the data")
```
