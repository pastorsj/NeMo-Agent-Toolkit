## Sequential Executor
Here we can use the Python SDK to develop a sequential executor, then save it to a config.yaml and run it from there.

The sequential executor runs a list of functions in sequence, passing the output of one function to the next.


```python
import os
import sys

# Import the NeMo-Agent-Toolkit module
module_path = os.path.abspath('../../../src/')
if module_path not in sys.path:
    sys.path.insert(0, module_path)
```

```python
import logging

logger = logging.getLogger()
logger.setLevel(logging.DEBUG)
```

```python
from nat.control_flow.sequential_executor import SequentialExecutor
from nat.utils.sdk.nat_workflow import NatWorkflow
from nat_sequential_executor.register import DataAnalyzerTool
from nat_sequential_executor.register import ReportGeneratorTool
from nat_sequential_executor.register import TextProcessorTool

# Define the sequential processing tools
text_processor = TextProcessorTool(
    name="text_processor",
)
data_analyzer = DataAnalyzerTool(
    name="data_analyzer",
)
report_generator = ReportGeneratorTool(
    name="report_generator",
)

# Create the sequential executor workflow
workflow = SequentialExecutor(
    tools=[text_processor, data_analyzer, report_generator],
    raise_type_incompatibility=False,
)

nat_workflow = NatWorkflow(
    entrypoint=workflow,
)
```

```python
import os
from pathlib import Path

path_to_yaml = Path(os.getcwd(), "config", "config.yaml").resolve()

# Create the config directory if it doesn't exist
if not path_to_yaml.parent.exists():
    os.makedirs(path_to_yaml.parent)

# Save the workflow to a config file
nat_workflow.save_to_config_file(path_to_yaml)

# Print out the config file content
with open(path_to_yaml) as f:
    print(f.read())
```

```python
# Test the sequential executor with sample text
sample_text = """
This is a sample text for analysis. It contains multiple sentences with varying complexity.
The sequential executor will process this text through three stages:
first cleaning and extracting statistics, then analyzing the content, and finally generating a report.
This demonstrates how functions can be chained together in a pipeline.
"""

await nat_workflow.prompt(sample_text)
```
